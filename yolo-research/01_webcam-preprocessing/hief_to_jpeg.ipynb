```python
from pathlib import Path

from PIL import Image
from pillow_heif import register_heif_opener


register_heif_opener()


def convert_folder(
    input_dir: str,
    output_dir: str,
    quality: int = 95,
) -> None:
    input_folder = Path(input_dir)
    output_folder = Path(output_dir)

    if not input_folder.exists():
        raise FileNotFoundError(f"입력 폴더가 없습니다: {input_folder}")

    output_folder.mkdir(parents=True, exist_ok=True)

    extensions = {".heic", ".heif"}

    heic_files = [
        file
        for file in input_folder.iterdir()
        if file.is_file() and file.suffix.lower() in extensions
    ]

    if not heic_files:
        print("변환할 HEIC/HEIF 파일이 없습니다.")
        return

    success_count = 0

    for input_file in heic_files:
        output_file = output_folder / f"{input_file.stem}.jpg"

        try:
            with Image.open(input_file) as image:
                rgb_image = image.convert("RGB")
                rgb_image.save(
                    output_file,
                    format="JPEG",
                    quality=quality,
                    optimize=True,
                )

            success_count += 1
            print(f"[완료] {input_file.name} → {output_file.name}")

        except Exception as error:
            print(f"[실패] {input_file.name}: {error}")

    print(f"\n총 {len(heic_files)}개 중 {success_count}개 변환 완료")


if __name__ == "__main__":
    convert_folder(
        input_dir="hief",
        output_dir="data",
    )
```
